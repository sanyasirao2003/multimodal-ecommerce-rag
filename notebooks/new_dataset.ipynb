```python
import pandas as pd

df = pd.read_csv("data/processed/products_clean_v1.csv")
print(df.columns)
```

```python
df["search_text"] = (
    df["title"].fillna("") + " " +
    df["category_clean"].fillna("") + " " +
    df["brand"].fillna("") + " " +
    df["description"].fillna("") + " " +
    df["color"].fillna("") + " " +
    df["gender"].fillna("") + " " +
    df["price_range"].fillna("") + " " +
    df["tags"].fillna("")
)
```

```python
def extract_tags(text):
    text = str(text).lower()
    tags = []

    keywords = ["running", "formal", "casual", "sports", "leather", "cotton"]
    for k in keywords:
        if k in text:
            tags.append(k)

    return " ".join(tags)

df["tags"] = (
    df["title"].fillna("") + " " +
    df["description"].fillna("") + " " +
    df["category_clean"].fillna("") + " " +
    df["brand"].fillna("") + " " +
    df["color"].fillna("")
).apply(extract_tags)
```

```python
def clean_text(text):
    return str(text).lower().strip()

df = df.applymap(clean_text)
```
